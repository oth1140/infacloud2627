```python
import requests
import datetime
import json
import sys

now = datetime.datetime.now()
print(now)

r = requests.get("https://api.ipify.org?format=json", timeout=10)
print("Status Code:", r.status_code)

if r.status_code != 200:
    print("Error:", r.status_code, r.text, file=sys.stderr)
    sys.exit(1)

ip_json = r.json()
print(ip_json)

json_str = json.dumps(ip_json)
print(json_str)

ip = ip_json.get("ip")

if ip is None:
    print("Unexpected response:", ip_json, file=sys.stderr)
    sys.exit(1)

print("IP Address:", ip)
```
