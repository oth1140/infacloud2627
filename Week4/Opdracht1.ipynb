```python
from netmiko import ConnectHandler

device = {
    "device_type": "cisco_ios",
    "host": "10.10.20.48",
    "port": 22,
    "username": "developer",
    "password": "C1sco12345",
    "conn_timeout": 30,
    "banner_timeout": 30,
}

try:
    connection = ConnectHandler(**device)
    connection.enable()
    output = connection.send_command("show ssh")
    print(output)
    connection.disconnect()
except Exception as e:
    print(f"Erreur : {e}")
```
