```python
from netmiko import ConnectHandler

v_device_type = "cisco_ios"
v_host = "10.10.20.48"
v_ssh_port = 22
v_username = "developer"
v_password = "C1sco12345"
v_secret = "C1sco12345"

device = {
    "device_type": v_device_type,
    "host": v_host,
    "port": v_ssh_port,
    "username": v_username,
    "password": v_password,
    "secret": v_secret,
    "conn_timeout": 30,
    "banner_timeout": 30,
}

connection = ConnectHandler(**device)
connection.enable()

config_commands = [
    "service password-encryption",
    "banner motd #Unauthorized access is prohibited by Netmiko#"
]

output = connection.send_config_set(config_commands)
print(output)

connection.save_config()

verify_output = connection.send_command("show running-config | include banner")
print("Verification result:")
print(verify_output)

connection.disconnect()
```
