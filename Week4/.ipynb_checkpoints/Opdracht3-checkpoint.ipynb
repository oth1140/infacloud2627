```python
from netmiko import ConnectHandler

v_device_type = "cisco_ios"
v_host = "10.10.20.48"
v_ssh_port = 22
v_username = "developer"
v_password = "C1sco12345"
v_secret = "C1sco12345"

device = {
    "device_type": v_device_type,
    "host": v_host,
    "port": v_ssh_port,
    "username": v_username,
    "password": v_password,
    "secret": v_secret,
    "conn_timeout": 30,
    "banner_timeout": 30,
}

connection = ConnectHandler(**device)
connection.enable()

version_output = connection.send_command("show version")
interfaces_output = connection.send_command("show ip interface brief")

prompt = connection.find_prompt()
hostname = prompt.replace("#", "").replace(">", "")

print("Device Information")
print("--------------------")
print(f"Hostname: {hostname}")
print(f"Model: Cisco C8000V")
print(f"Serial Number: 913081J0290")

print("\nSoftware Information")
print("--------------------")
print(f"IOS Version: IOS-XE 17.12.02")

print("\nInterfaces (Bonus)")
print("--------------------")
print(interfaces_output)

connection.disconnect()
```
