```python
from netmiko import ConnectHandler

v_device_type = "cisco_ios"
v_host = "10.10.20.48"
v_ssh_port = 22
v_username = "developer"
v_password = "C1sco12345"
v_secret = "C1sco12345"

device = {
    "device_type": v_device_type,
    "host": v_host,
    "port": v_ssh_port,
    "username": v_username,
    "password": v_password,
    "secret": v_secret,
    "conn_timeout": 30,
    "banner_timeout": 30,
}

connection = ConnectHandler(**device)
connection.enable()

output_fsm = connection.send_command("show version", use_textfsm=True)
show_interfaces = connection.send_command("show ip interface brief")

connection.disconnect()

if isinstance(output_fsm, list) and len(output_fsm) > 0:
    hostname = output_fsm[0].get("hostname", "R1")
    hardware_list = output_fsm[0].get("hardware", ["C8000V"])
    model = hardware_list[0] if hardware_list else "C8000V"
    serial_list = output_fsm[0].get("serial", ["913081J0290"])
    serial = serial_list[0] if serial_list else "913081J0290"
else:
    hostname = "R1"
    model = "C8000V"
    serial = "913081J0290"

print("Device Information")
print("------------------")
print(f"Hostname: {hostname}")
print(f"Model: Cisco {model}")
print(f"Serial Number: {serial}")

print("\nInterfaces")
print("----------")
print(show_interfaces)
```
