```python
%run filter_sysinfo.ipynb
import json

data = filter_info(collect_system_info())           
path = f"sysinfo_{data['computernaam']}.json"

json_text = json.dumps(data, indent=2, ensure_ascii=False)
with open(path, "w", encoding="utf-8") as f:
    f.write(json_text)

with open(path, encoding="utf-8") as f:
    reloaded = json.load(f)                

print("Types:", type(data), type(json_text), type(reloaded))
print("Identiek na inlezen:", reloaded == data)
for k in ("computernaam", "ram_totaal_gb", "ipv4_adres"):
    print(k, "=", data[k], "|", reloaded[k], "|", data[k] == reloaded[k])
print("Opgeslagen:", path)

##Een dict is iets dat Python in het geheugen bewaart. JSON is gewoon tekst, die je kan opslaan of doorsturen naar een ander programma. In JSON staan de namen altijd tussen dubbele aanhalingstekens, en True en None worden true en null.
```
